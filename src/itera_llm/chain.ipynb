### SimpleSequentialChain

```python
chat = criar_modelo()

prompt = PromptTemplate.from_template(
  """
    Escolha o molhor nome para im sobre uma empresa que desenvolve uma solução em {produto}
  """)

chain_nome = LLMChain(llm=chat, prompt=prompt)

prompt_descricao = PromptTemplate.from_template(
  """
  Sobre a empresa com o nome {nome_empresa}
  informe uma descrição de até 20 palavras
  """)

chain_descricao = LLMChain(llm=chat, prompt=prompt_descricao)

chain = SimpleSequentialChain(
  chains=[chain_nome, chain_descricao],
  verbose=True
)

chain.run(input=produto)
```

### SequentialChain

```python
chat = criar_modelo()

prompt = PromptTemplate.from_template(
  """
    Escolha o molhor nome para im sobre uma empresa que desenvolve uma solução em {produto}
  """)

chain_nome = LLMChain(llm=chat, prompt=prompt, output_key="nome_empresa")

prompt_descricao = PromptTemplate.from_template(
  """
  Sobre a empresa com o nome {nome_empresa} e que
  produz seguinte {produto}
  informe uma descrição de até 20 palavras de atividades dessa empresa
  """)

chain_descricao = LLMChain(llm=chat, prompt=prompt_descricao, output_key="descricao_empresa")

prompt_traducao = PromptTemplate.from_template(
  """
  Crie em um nome Italiano para a empresa de nome {nome_empresa}
  que possui a seguinte descrição {descricao_empresa}
  """
)

chain_traducao = LLMChain(llm=chat, prompt=prompt_traducao, output_key="nome_italiano")
```

```python
chain = SequentialChain(
  chains=[chain_nome, chain_descricao, chain_traducao],
  input_variables=["produto"],
  output_variables=["nome_empresa","descricao_empresa", "nome_italiano"],
  verbose=True
)

resposta = chain.invoke({"produto": produto})
print("Nome:", resposta["nome_empresa"])
print("Descrição:", resposta["descricao_empresa"])
print("Nome italiano:", resposta["nome_italiano"])
```
