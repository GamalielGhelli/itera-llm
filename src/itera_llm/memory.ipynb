```python
from langchain_classic.memory import ConversationBufferMemory

memory = ConversationBufferMemory(return_messages=True)
memory.chat_memory.add_user_message("Oá")
memory.chat_memory.add_ai_message("Como vai?")
```

```python
memory.load_memory_variables({})
```

```python
from langchain_openai.chat_models import ChatOpenAI
from langchain_classic.chains.conversation.base import ConversationChain


chat = ChatOpenAI()
memory = ConversationBufferMemory()
conversation = ConversationChain(
  llm = chat,
  memory=memory,
  verbose=True
)
```

```python
conversation.predict(input="Qual é o meu nome?")
```

```python
from langchain_classic.memory import ConversationTokenBufferMemory

chat = ChatOpenAI()
memory = ConversationTokenBufferMemory(llm=chat, max_token_limit=150)
conversation = ConversationChain(
  llm = chat,
  memory=memory,
  verbose=True
)
```

```python
conversation.predict(input="faça uma poesia de 150 palavras")
```

```python
conversation.predict(input="sobre o que é a poesia?")
```
